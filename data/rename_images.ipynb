```python
import pandas as pd
import numpy as np
from glob import glob
import os
```

```python
img_abs_path = "/home/csikasote/Documents/IciBemba/igc-mu-cibemba/data/image"
```

```python
def check_for_unique_images(img_abs_path):
    images_list = glob(f"{img_abs_path}/*.jpg")
    unique_image_list = []
    count = 0
    for filename in images_list:
        image_id = os.path.basename(filename)
        if '_' in image_id:
            count = count + 1
            print(count, image_id)
            first_part = image_id.split('_')[0]
            dst =f"{img_abs_path}/{first_part}.jpg"
            if os.path.exists(dst):
                print(dst, 'EXISTS')
                os.remove(count, filename)
            else:
                os.rename(filename, dst)
```

```python
check_for_unique_images(img_abs_path)
```
